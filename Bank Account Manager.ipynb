Project Title: Bank Account Manager
Scenario:
A bank wants to track account balances and transactions for each customer in an object-orientated way.

```python
# Create a class BankAccount with account_holder and balance attributes.
# Implement methods for deposit(), withdraw(), and display_balance().
# Use both local and global variables appropriately, such as a global transaction counter.

# Global variable

transactions = 0

class BankAccount:
    def __init__(self,acc_hol,bal=0):
        self.account_holder = acc_hol
        self.balance = bal

    # Implement methods for deposit(), withdraw(), and display_balance().
    def deposit(self,credit_amount):
        global transactions
        self.balance += credit_amount
        transactions += 1
        print(f'{credit_amount} deposited successfully.')

    def withdraw(self,debit_amount):
        global transactions
        # Prevent withdrawals that would result in a negative balance.
        if debit_amount > self.balance:
            return 'Insufficient Balance'
        self.balance -= debit_amount
        transactions += 1
        print(f'{debit_amount} withdrawn successfully.')
    def display_balance(self):
        return f'Hi {self.account_holder} Your Available Balance: {self.balance}'

# Demonstrate calling methods using both positional and keyword arguments.

acc1 = BankAccount("Maha",2000)
print(acc1.display_balance())

acc1.deposit(2000)
print(acc1.display_balance())

acc1.withdraw(2000)
print(acc1.display_balance())

print(acc1.withdraw(2500))
```

```python
acc2 = BankAccount(bal=200,acc_hol="Saranya")
print(acc2.display_balance())

acc2.deposit(5000)

print(f'Total Transactions:{transactions}')
```
