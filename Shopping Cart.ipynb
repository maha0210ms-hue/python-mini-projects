Project Title: Shopping Cart System

You are creating a simple billing system where customers can enter the prices of items they want to purchase. The program should keep accepting prices until the user decides to stop, then apply appropriate discounts before generating the final bill.

```python
sub_total = 0
total_items = 0
discount_amount = 0
final_amount = 0

# Continuously accept item prices using a while loop.
# Stop taking input when the user enters 0 (sentinel value).
# Keep track of the total number of items and the subtotal amount.
# Ensure no negative prices are allowed (ask again without counting as an item).

while True:
    price = float(input("Enter the price:"))
    if price < 0:
        continue
    elif price == 0:
        break
    sub_total += price
    total_items += 1

# Apply discounts based on the subtotal:
# 20% for totals ≥ 1000
# 10% for totals ≥ 500
# 5% for totals ≥ 200
# 0% for totals < 200
    
if sub_total >= 1000:
    discount_amount = 0.2 * sub_total
elif sub_total >= 500:
    discount_amount = 0.1 * sub_total
elif sub_total >= 200:
    discount_amount = 0.05 * sub_total
else:
    discount_amount = 0 * sub_total
    
   
final_amount = sub_total - discount_amount
    
print("BILL SUMMARY".center(50, "-"))
print(f'Total Items = {total_items}')
print(f'Total Amount = {sub_total:.2f}')
print(f'Discounted Amount = {discount_amount:.2f}')
print(f'Amount to be paid = {round(final_amount)}')
print("-" * 50)

```
