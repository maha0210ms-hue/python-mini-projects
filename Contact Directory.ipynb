Project Title: Contact Directory
Scenario:
You have been assigned to build a contact management system where employees can store and retrieve contact details efficiently.

```python
# Create a 2D list where each row holds a contact’s name, phone number, and email address.

contact_management_system = [
    ['Name','Phone Number','Email'],
    ['Anandhi',9999999999,'anandhi@gmail.com'],
    ['Akshaya',7777777777,'akshaya@yahoo.com']
                   ]
contact_management_system
```

```python
# Allow the user to add a new contact.

contact_management_system.append(['Bala',8888888888,'bala@gmail.com'])
contact_management_system
```

```python
# Remove the last item

contact_management_system.pop()
```

```python
# Get new items from the users using input() and add those items to the list

Name = input('Enter your name:')
phone_number = int(input('Enter your mobile number:'))
email = input('Enter your email:')

contact_management_system.append([Name,phone_number,email])
```

```python
contact_management_system
```

```python
#  Sort the list in ascending order based on any values
contact_management_system.append(['Bala',8888888888,'bala@gmail.com'])

header = contact_management_system[0]
data = contact_management_system[1:]
data.sort(key = lambda a :a[0])
print(data)

contact_management_system = [header] + data

contact_management_system
```
