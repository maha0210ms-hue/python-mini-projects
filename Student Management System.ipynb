Project Title: Student Management System
Scenario:
 You are tasked with building a Student Management System that integrates multiple Python concepts to help staff manage students' data, performance, and reporting.

```python
# Use dictionaries and lists to store and retrieve student information such as name, age, and grades.

student_details = {
    "name":['Anu','Deepan','Dev','Madhu','Saranya'],
    "age":[25,22,24,27,28],
    "grades":['A','B','C','B','A']
}

print(student_details['name'])
print(student_details['age'])
print(student_details['grades'])
```

```python
# Allow users to add, delete, update, and search for students.

name = input("Enter student's name:")
age = int(input("Enter student's age:"))
grades = input("Enter student's grade")

student_details['name'].append(name)
student_details['age'].append(age)
student_details['grades'].append(grades)
```

```python
print(student_details)
```

```python
# deleting student

index = student_details['name'].index('Dev')
student_details['name'].pop(index)
student_details['age'].pop(index)
student_details['grades'].pop(index)

print(student_details)
```

```python
# updating student details

index = student_details['name'].index('Deepan')

student_details['grades'][index] = 'B+'

print(student_details)
```

```python
# searching student details

name = 'Madhu'
if name in student_details['name']:
    index = student_details['name'].index(name)

    print(student_details['name'][index])
    print(student_details['age'][index])
    print(student_details['grades'][index])
```

```python
# Track grades and categorize performance using functions and conditionals.

for name,grade in zip(student_details['name'],student_details['grades']):
    
    if grade == 'A+':
        print(f'{name} is a Top performer')
    elif grade == 'A':
        print(f'{name} is a Good performer')
    elif grade == 'B+':
        print(f'{name} is an Above Average performer')
    elif grade == 'B':
        print(f'{name} is an Average performer')
    else:
        print(f'{name} needs improvement')
```

```python
# Optionally implement file handling to save and load student data between program runs.

import json
# save data to json file
with open('students.json', 'w') as file:
    json.dump(student_details, file)

print("Student data saved successfully")
```

```python
# load data from json file
with open('students.json', 'r') as file:
    loaded_data = json.load(file)

print("\nLoaded Student Data")
print(loaded_data)
```

```python
# Use exception handling to manage invalid inputs.
try:

    student_name = input("\nEnter student name: ")

    student_age = int(input("Enter student age: "))

    student_grade = input("Enter student grade: ")

    student_details['name'].append(student_name)
    student_details['age'].append(student_age)
    student_details['grades'].append(student_grade)

    print("\nStudent added successfully")

except ValueError:
    print("Invalid input! Age must be a number.")

except Exception as e:
    print("Error:", e)
    
```

```python
# Implement sorting and filtering of students for reports.

sorted_student = sorted(zip(student_details['name'],
                            student_details['age'],
                            student_details['grades']),
                        key = lambda x : x[0],
                        reverse = True)
print("Sorted Student by name in descending order")
for name,age,grade in sorted_student:
    print(name,age,grade)
   
    
```

```python
good_performers = []
for name,grade in zip(student_details['name'],student_details['grades']):
    if grade == 'A':
        good_performers.append(name)
if len(good_performers) > 1:   
    print(f'{",".join(good_performers)} are good performers') 
elif len(good_performers) == 1:
    print(f'{(good_performers[0])} is a good performer')
else:
    print('No good performers')
    
```
