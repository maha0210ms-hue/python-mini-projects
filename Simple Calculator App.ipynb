Project Title: Simple Calculator App
Scenario:
A company wants to provide users with a simple calculator to perform basic mathematical operations.

```python
# Create functions for addition, subtraction, multiplication, and division.
# Ask the user to input two numbers and the operation they want to perform.
# Call the appropriate function and display the result.
# Use built-in functions like abs(), round(), and pow() within your solution.
# Handle division by zero gracefully and inform the user.



def addition(a,b):
    return a + b
def subtraction(a,b):
    return a - b
def multiplication(a,b):
    return a * b
def division(a,b):
    return a / b

while True:

    num1 = float(input("Enter first number : "))
    num2 = float(input("Enter second number : "))
    operation = input("Enter the operator to perform calculation (+,-,*,/) : ")

    if operation == '+':
        result = addition(num1,num2)
        print(f'{num1} + {num2} = {round(result,2)}')
        
    elif operation == '-':
        if num2 > num1:
            result = abs(subtraction(num1,num2))
            print(f'Absolute result of {num1} - {num2} = {round(result,2)}')
        else:
            result = subtraction(num1,num2)
            print(f'{num1} - {num2} = {round(result,2)}')
        
    elif operation == '*':
        result = multiplication(num1,num2)
        print(f'{num1} * {num2} = {round(result,2)}')
        
    elif operation == '/':
        if num2 != 0:
            result = division(num1,num2)
            print(f'{num1} / {num2} = {round(result,2)}')
            
        else:
            print("cannot divide a number by zero")
            continue
    else:
        print("Enter a valid operator")
        continue
    num = int(input("Enter power ="))
    res = pow(result,num)
    print(f'Result ** {num} = {round(res,2)}')
    break
    
```
